```python
import pandas as pd
age=pd.Series([1,2,3,4,5])
print(age)
```

```python
age=([10,20,30,40])
series=pd.Series(age,index=["sri","mahesh","karthi","manish"])
print(series)
```

```python
age={
    "sri"     : 21,
    "mahesh"  : 22,
    "karthi"  : 23,
    "manish"  : 26
}
series=pd.Series(age)
print(series)
```

```python
age=pd.Series([1,2,3,4,5])
print(age)
print(age.iloc[2])
```

```python
age=([10,20,30,40])
series=pd.Series(age,index=["sri","mahesh","karthi","manish"])
print(series)
print(series.loc["mahesh"])
```

```python
sales=pd.Series([2000,3000,7000,8000],index=["Q1","Q2","Q3","Q4"])
print(sales[["Q1","Q4"]])
```

```python
sales1 = pd.Series([100,200,300],index=['Q1','Q2','Q3'])
sales2 = pd.Series([400,500,600],index=['Q1','Q2','Q3'])
print("Addition:\n",sales1 + sales2)
print("Subraction:\n",sales1 - sales2)
print("Multiplication:\n",sales1 * sales2)
print("Division:\n",sales1 / sales2)
```
